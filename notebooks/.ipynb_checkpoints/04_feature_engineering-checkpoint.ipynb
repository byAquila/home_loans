# 04. Feature Engineering - Home Loan Approval Prediction

**Objective
Create new useful features and prepare the data for modeling.

**Main Goals:
-Create TotalIncome.
-Create EMI-related features.
-Encode categorical variables.
-Prepare final datasets for modeling.


```python
import pandas as pd
import numpy as np

train = pd.read_csv('../data/processed/train_cleaned.csv')
test =pd.read_csv('../data/processed/test_cleaned.csv')

print("Train shape:", train.shape)
print("Test shape:", test.shape)
```

```python
# 1. Total Income
train['TotalIncome'] = train['ApplicantIncome'] + train['CoapplicantIncome']
test['TotalIncome'] = test['ApplicantIncome'] + test['CoapplicantIncome']

# 2. EMI = Loan Amount/ Loan_Amount_Term
train['EMI'] = train['LoanAmount']/ train['Loan_Amount_Term']
test['EMI'] = test['LoanAmount']/ test['Loan_Amount_Term']

# 3. Income to Loan Ratio
train['Income_to_Loan'] = train['TotalIncome']/ train['LoanAmount']
test['Income_to_Loan'] = test['TotalIncome']/ test['LoanAmount']

print("New features created: TotalIncome, EMI, Income_to_Loan")
print(train[['TotalIncome', 'EMI', 'Income_to_Loan']].head())
```

```python
from sklearn.preprocessing import LabelEncoder

# Columns to encode
cat_cols = ['Gender', 'Married', 'Education', 'Self_Employed', 'Property_Area']

le = LabelEncoder()

for col in cat_cols:
    # Fit on train and transform both train and test
    train[col] = le.fit_transform(train[col])
    test[col] = le.transform(test[col])
    print(f"{col} encoded successfully")

print("\nEncoding completed.")
```

```python
# Feaures for modeling
features = ['Gender', 'Married', 'Dependents', 'Education', 'Self_Employed', 'ApplicantIncome', 'CoapplicantIncome', 'LoanAmount', 'Loan_Amount_Term', 'Credit_History', 'Property_Area', 'TotalIncome', 'EMI', 'Income_to_Loan']
x= train[features]
y= train['Loan_Status'].map({'Y':1, 'N':0})

X_test = test[features]

print("x shape:", x.shape)
print("y shape:", y.shape)
print("X_test shape:", X_test.shape)
```

```python
# Saving the engineered data
train_final = x.copy()
train_final['Loan_Status'] = y

train_final.to_csv('../data/processed/train_final.csv', index=False)
X_test.to_csv('../data/processed/test_final.csv', index=False)

print("Final datasets saved.")
```
