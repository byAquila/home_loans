# 02. Data Cleaning - Home Loan Approval Prediction

**Objective
Clean the dataset so it is ready for analysis and modeling.

**Main Goals This Notebook:
-Handle missing values.
-Fix data type and inconsistent values.
-Treat outliers where necessary
-Create cleaned versions of train and test sets.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
sns.set_style('whitegrid')

train = pd.read_csv('../data/raw/train.csv')
test = pd.read_csv('../data/raw/test.csv')

print("Train shape:", train.shape)
print("Test shape:", test.shape)
print("Data loaded successfully.")
```

```python
print("Missing values in Train before cleaning:")
print(train.isnull().sum().sort_values(ascending=False))
```

```python
# Fill missing categorical values with mode
for col in ['Gender', 'Married', 'Dependents', 'Self_Employed', 'Credit_History']:
    mode_value = train[col].mode()[0]
    train[col] = train[col].fillna(mode_value)
    test[col] = test[col].fillna(mode_value)
    print(f"{col} filled with: {mode_value}")
```

```python
# LoanAmount - fill with median
median_loan = train['LoanAmount'].median()
train['LoanAmount'] = train['LoanAmount'].fillna(median_loan)
test['LoanAmount'] = test['LoanAmount'].fillna(median_loan)
print(f"LoanAmount filled with median: {median_loan}")

#Loan_Amount_Term - fill with mode
mode_term = train['Loan_Amount_Term'].mode()[0]
train['Loan_Amount_Term'] = train['Loan_Amount_Term'].fillna(mode_term)
test['Loan_Amount_Term'] = test['Loan_Amount_Term'].fillna(mode_term)
print(f"Loan_Amount_Term filled wth mode: {mode_term}")
```

```python
train['Dependents'] = train['Dependents'].replace('3+', '3')
test['Dependents'] = test['Dependents'].replace('3+', '3')

# Convert to integer
train['Dependents'] = train['Dependents'].astype(int)
test['Dependents'] = test['Dependents'].astype(int)

print("Dependents fixed and converted to integer.")
print(train['Dependents'].value_counts())
    
```

```python
print("Missing values in Train after cleaning:")
print(train.isnull().sum())

print("\nMissing values in Test after cleaning:")
print(test.isnull().sum())
```

```python
# Create the processed folder if it doesn't exist (optional safety)
import os
os.makedirs('../data/processed', exist_ok=True)

# Save cleaned files
train.to_csv('../data/processed/train_cleaned.csv', index=False)
test.to_csv('../data/processed/test_cleaned.csv', index=False)

print("Cleaned data saved successfullyy!")
print("Files saved in: data/processed/")
```

### Summary of Data Cleaning

**Actions taken
-Filled missing categorical vallues ('Gender', 'Married', 'Dependents', Self_Employed', 'Credit_History') with the mode.
-Filled 'LoanAmount' with the median
-Filled 'Loan_Amount_Term' with the mode.
-Replaced '3+' in 'Dependents' with '3' and converted the column to integer.

**Result:
-No missing values remain in Train or test sets
-Cleaned datasets saved as:
-- 'data/processed/train_cleaned.csv'
--'data/processed/test_cleaned.csv'
